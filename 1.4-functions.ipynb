```python
def conv_temp(temp , unit):
    if unit == "C":
        return temp *9/5 +32  # celsius to Fahrenheit
    
    elif unit =="F":
        return (temp - 32) * 5/9  # Fahrenheit to celsius

    else:
        return None


print(conv_temp(25,'C'))
print(conv_temp(187,"F"))
     
```

```python
def strg_pass(password):
    """ This function check if the password is strong or not """ 

    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char in "!@#$%^&_*()"  for char in password):
        return False
    return True



#calling the function 
print(strg_pass("Weak_pwd"))
print(strg_pass("Strong_pwd@123"))
```

```python
#Calculate the total cost of items in a shopping cart

def cal_total_cost(cart):
    total_cost = 0 
    for item in cart:
        total_cost += item["price"]*item["quantity"]
    return total_cost






#Example cart data

cart = [
    {'name':'Apple','price':0.5,'quantity':4},
    {'name':'Banana','price':0.3,'quantity':6},
    {'name':'Orange','price':0.7,'quantity':3}
]

total_cost = cal_total_cost(cart)
print(total_cost)
```

```python
#Checking if the given string is the palindrome or not

#eg :- aba 
def palindrome_or_not(s):
    s=s.lower().replace(" ","")

    return s ==s[::-1]

print(palindrome_or_not("A man a plan a canal Panama"))
print(palindrome_or_not("Hello"))
```

```python
#Calculate the factorials of a number using recursion

def factorial(n):
    if n ==0: 
        return 1 
    else :
        return n * factorial(n-1)
    
print(factorial(4))
    

```

Lambda function in python

lambda functions are small anonymous fuctions defined using the lambda keyword. They can have any number of arguments but only one expression. They are commonly used for short operations or as arguments to higher-order functions.

```python
# normal function
def addition(a,b):
    return a+b 
```

```python
#lambda function
addtion = lambda a,b:a+b
print(addtion(2,3))
```

```python
even = lambda n:n%2==0
print(even(13)) 
```

```python
addtion1 = lambda a,b,c:a+b+c
print(addtion1(2,3,4))
```

```python
#default function
def square(number):
    result = []
    for i in number: 
        result.append(i**2)
    return result
        

number = [1,2,3,4,5]

print(square(number))
```

```python
#map() applies the fuction to all item in a list
number=[1,2,3,4,5]
list(map(lambda x:x**2,number))
```

The map() function in python

The map() function applies a given function to all items in an input list (or any other iterable) and returns a mpa object (an iterator), this is particularly/useful for transforming data in a list comprehensively.

```python
# map fuction want only two things
# map(parameter ("function"),iterable)

#eg
def square(x):
    return x*x 

numbers =[1,2,3,4,5,6]
list(map(square,numbers))
```

```python
#Map multiple iterables

num1 = [1,2,3]
num2 = [4,5,6]

add_nums = list(map(lambda x , y : x+y , num1,num2))
print(add_nums)
```

```python
# map() to convert a list of string to int

str_num = ['1','2','3','4']
int_number = list(map(int, str_num))
print(int_number)
```

```python
words=['apple','banana','cherry']
upper_word=list(map(str.upper,words))
print(upper_word)
```

The filter() function in python

The filter function constructs an iterator from elements of an iterable for which  a fuctions return true. it is used to filter out items from a list (or any other iterable) based on a condition

```python
def even(num):
    return num%2==0

even(24)
```

```python
num3 = [2,3,4,5,7,12,43,22,18]
list(filter(even,num3))# filter(function , list (iterator))
```

```python
#filter with a lambda function
numbers= [1,2,3,4,5,6,7,8]
greater_than_five = list(filter(lambda x:x>5 , numbers))
print(greater_than_five)
```

```python
numbers= [1,2,3,4,5,6,7,8]
greater_than_five = list(filter(lambda x:x>5 and x%2==0, numbers))
print(greater_than_five)
```
