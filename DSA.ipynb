```python
#Amstrong number :
num = 153
n = num ; count = 0 ; AN = 0
while n>0:
  last_digit = n % 10
  count = count + 1
  n = n // 10
print('Count :' , count)

n = num
while n > 0 :
  ld = n % 10
  AN = AN + ld**count
  n = n //10
print('AN',AN)

if (num == AN) : print(f"it's an amstrong number : {num} == {AN}")
else : print(f'it is not a amstrong number : {num} == {AN}')


```

```python
# factorial of the given number :
num = int(input("Enter the number : "))
n = num ; fact = 1
while n > 0 :
  fact = fact * n
  n = n-1
print(f"Factorial of {num} : {fact}")
```

```python
num = int(input('Enter the number : '))
factors = []
for i in range(1,num+1):
  if num % i == 0 : factors.append(i)
print(f'the factor of{num} : {factors}')

#TC = O(N)
```

```python
#optimal solution
num = int(input("enter the number :"))
#code to find the square root of the given number
def square_root(num):
  for i in range(1,num+1):
    if (i*i == num) :
      print(i)
      break
result = []
for i in range(1 , square_root(num) +1):
  if num % i == 0 :
    if num // i != i : result.append(num//i)
print(result)
```

```python
# store frequency in the frequency
nums = [5,6,7,7,1,9,11,1,1,5,7,8]                          # TC = O(N) , SC =O(n)
frequency_map = {}

for i in range(0 , len(nums)):
  if nums[i] in frequency_map :
    frequency_map[nums[i]] +=1
  else :
     frequency_map[nums[i]] = 1

print(frequency_map)
```

```python
#Method 2
nums = [5,6,7,7,1,9,11,1,1,5,7,8]
n = len(nums)
hash_map = {}
for i in range(n):
  hash_map[nums[i]] = hash_map.get(nums[i],0)+1
print(hash_map)
```
